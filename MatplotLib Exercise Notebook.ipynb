## Exercise 2 - Import Matplotlib using

import matplotlib.pyplot as plt



```python
import matplotlib.pyplot as plt
import pandas as pd

car_data = pd.read_csv("car data.csv")
car_data.head()
```

## Exercise 3 - Draw Barchart




```python
fuel_counts = car_data["Fuel_Type"].value_counts()
print(fuel_counts)
plt.bar(fuel_counts.index, fuel_counts.values)
plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.show()
```

## Exercise 4 -  Draw Histogram


```python

selling_price_counts_sorted = car_data.sort_values(by="Selling_Price", ascending=False)
print(selling_price_counts_sorted)

selling_price_counts = car_data["Selling_Price"].value_counts()
plt.hist(car_data["Selling_Price"])
plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Frequency")
plt.show()
```

## Exercise 5 -  Draw Piechart



```python
transmission_counts = car_data["Transmission"].value_counts()
print(transmission_counts)
plt.pie(transmission_counts.values, labels=transmission_counts.index)
plt.title("Cars by Transmission Type")
plt.show()
```
