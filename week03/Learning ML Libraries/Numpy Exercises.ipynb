# In-Class Exercise 3: Slice an Array

```python
arr_ex3 = np.array([2,4,6,8,10,12,14,16])
print("Slicing arr_ex3 with starting index 2 and ending index 6:", arr_ex3[2:6])
```

###### Starting index: 2
###### Ending index: 6
###### Slice: arr_ex3[2:6]
###### Why isn't index 6 included? 
###### Because Numpy slicing includes the starting index but excludes the ending index.

# In-Class Exercise 4: Sort Numeric and Text Data

```python
arr_ex4 = np.array([45,12,78,3,30])
arr_ex4_sorted = np.sort(arr_ex4)
arr_ex4_str = np.array(["orange","apple","banana","kiwi"])
arr_ex4_str_sorted = np.sort(arr_ex4_str)
print(arr_ex4_sorted)
print(arr_ex4_str_sorted)
```
