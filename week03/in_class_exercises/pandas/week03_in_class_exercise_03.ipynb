```python
import pandas as pd

player_data = [
    ['Player 1', 32, 75, 5428000], ['Player 2', 33, 74, 3428000],
    ['Player 3', 34, 70, 8428000], ['Player 4', 35, 80, 4428000],
    ['Player 5', 36, 100, 4528000], ['Player 6', 38, 72, 7028000],
    ['Player 7', 22, 85, 2528000]
]
df = pd.DataFrame(player_data, columns = ['Name', 'Age', 'Weight', 'Salary'])
df
```

```python
df_first_four_rows = df.iloc[0:4]
df_first_four_rows
```

```python
# Select only Name, Age, and Salary
df_Name_Age_Salary_cols = df.iloc[:,[0,1,3]]
df_Name_Age_Salary_cols
```

```python
# Retrieve one specific salary using iloc
specific_salary = df.iloc[4,3]
print("Specific Salary:", specific_salary)
```

```python
# Filter players whose Age is greater than 34
players_older_than_34 = df[df['Age'] > 34]
print(players_older_than_34)
```

```python
# Create a second df with Name and Team
name_and_team = {
    'Name': ['Player 1','Player 2','Player 3','Player 4','Player 5','Player 6','Player 7'],
    'Team': ['Team A','Team B','Team A','Team C','Team B','Team C','Team A']
}

team_df = pd.DataFrame(name_and_team)
print(team_df)
```

```python
# Merge it with the player df
merged_df = pd.merge(df, team_df)
print(merged_df)
```

```python
# Export to csv with index=False
merged_df.to_csv('merged_df.csv', index=False)
```
